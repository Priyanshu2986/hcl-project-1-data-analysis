```python
# Import required libraries for data handling and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset from the CSV file
df = pd.read_csv('heart_failure_clinical_records_dataset.csv')

# Display the first 5 rows using head()
print("First 5 rows:")
print(df.head())

```

```python
# Display the last 5 rows using tail()
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv('heart_failure_clinical_records_dataset.csv')
print("\nLast 5 rows:")
print(df.tail())
```

```python
# Display the dataset dimensions using shape
print("\nDataset Shape (rows, columns):", df.shape)
```

```python
# Display dataset column names, non-null counts, and data types using info()
print("\nDataset Info:")
df.info()
```

```python
# Display summary statistics for all numerical columns using describe()
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Question 6: Check for missing values and duplicate records
print("\nMissing Values per Column:")
print(df.isnull().sum())
print("Duplicate Records Count:", df.duplicated().sum())
```

```python
# Question 1: Examine the age distribution using summary stats and age bins
print("\nQuestion 1 - Age Distribution Summary:")
print(df['age'].describe())
age_bins = pd.cut(df['age'], bins=[39, 50, 60, 70, 80, 100], labels=['40-50', '51-60', '61-70', '71-80', '81+'])
print("\nAge Group Counts:")
print(age_bins.value_counts().sort_index())
```

```python
# Question 2: Count records by DEATH_EVENT using value_counts()
print("\nQuestion 2 - Record Counts by DEATH_EVENT (0 = Survived, 1 = Deceased):")
death_counts = df['DEATH_EVENT'].value_counts()
print(death_counts)
```

```python
# Question 3: Calculate mean, median, minimum, and maximum for age, ejection_fraction, and serum_creatinine
print("\nQuestion 3 - Mean, Median, Min, and Max:")
summary_metrics = df[['age', 'ejection_fraction', 'serum_creatinine']].agg(['mean', 'median', 'min', 'max'])
print(summary_metrics.round(2))
```

```python
# Perform calculations using a NumPy array as required
age_array = df['age'].to_numpy()
ef_array = df['ejection_fraction'].to_numpy()
sc_array = df['serum_creatinine'].to_numpy()
print("\nNumPy Array Calculations:")
print(f"Age -> Mean: {np.mean(age_array):.2f}, Median: {np.median(age_array):.2f}, Min: {np.min(age_array):.2f}, Max: {np.max(age_array):.2f}")
print(f"Ejection Fraction -> Mean: {np.mean(ef_array):.2f}, Median: {np.median(ef_array):.2f}, Min: {np.min(ef_array)}, Max: {np.max(ef_array)}")
print(f"Serum Creatinine -> Mean: {np.mean(sc_array):.2f}, Median: {np.median(sc_array):.2f}, Min: {np.min(sc_array):.2f}, Max: {np.max(sc_array):.2f}")
```

```python
# Question 4: Compare average ejection_fraction and serum_creatinine across DEATH_EVENT groups using groupby()
print("\nQuestion 4 - Average Ejection Fraction & Serum Creatinine by DEATH_EVENT:")
grouped_comparison = df.groupby('DEATH_EVENT')[['ejection_fraction', 'serum_creatinine']].mean()
print(grouped_comparison.round(2))
```

```python
# Question 5: Count records with and without diabetes, high blood pressure, and anaemia
print("\nQuestion 5 - Counts for Diabetes, High Blood Pressure, and Anaemia (0 = No, 1 = Yes):")
for condition in ['diabetes', 'high_blood_pressure', 'anaemia']:
    print(f"\n{condition}:")
    print(df[condition].value_counts())
```

```python
# Demonstrate Pandas filtering and sorting on senior patients (age > 75) sorted by serum_creatinine
senior_filtered_sorted = df[df['age'] > 75].sort_values(by='serum_creatinine', ascending=False)
print("\nFiltered & Sorted Sample (Patients Aged > 75, Top 5 by Serum Creatinine):")
print(senior_filtered_sorted[['age', 'ejection_fraction', 'serum_creatinine', 'DEATH_EVENT']].head())
```

```python
# Chart 1: Create a bar chart for DEATH_EVENT counts
plt.figure(figsize=(6, 4))
death_counts.plot(kind='bar', color=['#2b5c8f', '#d9534f'], edgecolor='black')
plt.title('Record Counts by DEATH_EVENT')
plt.xlabel('DEATH_EVENT (0 = Survived, 1 = Deceased)')
plt.ylabel('Number of Records')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('death_event_bar_chart.png')
plt.show()
```

```python
# Chart 2: Create a histogram for the age distribution
plt.figure(figsize=(6, 4))
plt.hist(df['age'], bins=12, color='#4a90e2', edgecolor='black')
plt.title('Age Distribution of Patients')
plt.xlabel('Age (Years)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('age_histogram.png')
plt.show()
```
